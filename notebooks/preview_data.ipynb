# Hospital Price Data – Quick Preview
Shows the first 5 rows of each dataset in `data/`.

**Colab:** upload the `data/` folder to Drive, mount it, and set `DATA_DIR` below. **Local:** run as-is.

```python
import json
from pathlib import Path
import pandas as pd

# Colab: from google.colab import drive; drive.mount('/content/drive')
# DATA_DIR = Path('/content/drive/MyDrive/TrueCost/data')
DATA_DIR = Path('../data')
N = 5
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_colwidth', 40)
sorted(p.name for p in DATA_DIR.iterdir())

import csv

def read_csv_preview(path, n=N):
    """Auto-detect the real header row (hospital files have metadata rows above it)."""
    with open(path, newline='', encoding='utf-8', errors='replace') as f:
        rows = [r for _, r in zip(range(30), csv.reader(f))]
    counts = [sum(1 for c in r if c.strip()) for r in rows]
    header = counts.index(max(counts))
    print(f'{Path(path).name}: header found on line {header + 1}')
    return pd.read_csv(path, skiprows=header, nrows=n, low_memory=False, index_col=False, encoding_errors='replace')
```

## Mercy Hospital of Buffalo (CSV)

```python
read_csv_preview(DATA_DIR/'16-0756336_Mercy-Hospital-of-Buffalo_StandardCharges.csv')
```

## Roswell Park (JSON)

```python
with open(DATA_DIR/'161552370_roswell-park-cancer-institute_standardcharges.json', encoding='utf-8-sig') as f:
    data = json.load(f)
print('Top-level keys:', list(data.keys()) if isinstance(data, dict) else type(data))
for k, v in (data.items() if isinstance(data, dict) else []):
    if isinstance(v, list) and v and isinstance(v[0], dict):
        print(f"\n'{k}': {len(v)} records")
        display(pd.json_normalize(v[:N]))
```
